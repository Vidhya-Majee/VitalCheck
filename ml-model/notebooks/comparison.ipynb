```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
metrics = ['Accuracy', 'Precision', 'Recall', 'F1-Score']

diabetes_scores = {
    'Logistic Regression': [0.730, 0.341, 0.760, 0.471],
    'Random Forest': [0.841, 0.487, 0.196, 0.279],
    'XGBoost': [0.835, 0.438, 0.169, 0.244],
}

heart_scores = {
    'Logistic Regression': [0.753, 0.248, 0.797, 0.378],
    'Random Forest': [0.899, 0.367, 0.097, 0.153],
    'XGBoost': [0.752, 0.244, 0.782, 0.372],
}
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5.5))
x = np.arange(len(metrics))
width = 0.25
colors = ['#4A90D9', '#2E7D32', '#C62828']

for ax, (title, scores) in zip(axes, [('Diabetes Model Comparison', diabetes_scores),
                                        ('Heart Disease Model Comparison', heart_scores)]):
    for i, (model, vals) in enumerate(scores.items()):
        ax.bar(x + (i-1)*width, vals, width, label=model, color=colors[i])
    ax.set_xticks(x)
    ax.set_xticklabels(metrics)
    ax.set_ylim(0, 1)
    ax.set_title(title, fontsize=12, weight='bold')
    ax.legend(fontsize=8)
    ax.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.savefig('../saved_models/model_comparison_real.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
fig, axes = plt.subplots(2, 1, figsize=(9, 11))
x = np.arange(len(metrics))
width = 0.25
colors = ['#4A90D9', '#2E7D32', '#C62828']

for ax, (title, scores) in zip(axes, [('Diabetes Model Comparison', diabetes_scores),
                                        ('Heart Disease Model Comparison', heart_scores)]):
    for i, (model, vals) in enumerate(scores.items()):
        bars = ax.bar(x + (i-1)*width, vals, width, label=model, color=colors[i],
                       edgecolor='white', linewidth=0.7)
        ax.bar_label(bars, fmt='%.2f', fontsize=7.5, padding=2)
    ax.set_xticks(x)
    ax.set_xticklabels(metrics, fontsize=11)
    ax.set_ylim(0, 1.05)
    ax.set_ylabel('Score', fontsize=10)
    ax.set_title(title, fontsize=13, weight='bold', pad=12)
    ax.legend(fontsize=9, loc='upper right', framealpha=0.9)
    ax.grid(axis='y', linestyle='--', alpha=0.35)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout(pad=2.5)
plt.savefig('../saved_models/model_comparison_real2.png', dpi=200, bbox_inches='tight', facecolor='white')
plt.show()
```
