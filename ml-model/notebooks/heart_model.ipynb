```python
import pandas as pd
```

```python
df_heart=pd.read_csv("../data/heart.csv")
print("Shape:",df_heart.shape)
```

```python
df_heart.head()
```

```python
df_heart.info()
```

```python
print(df_heart['HeartDiseaseorAttack'].value_counts())
```

```python
corr = df_heart.corr()['HeartDiseaseorAttack'].sort_values(ascending=False)
print(corr)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X_h = df_heart.drop(columns=['HeartDiseaseorAttack'])
y_h = df_heart['HeartDiseaseorAttack']

X_h_train, X_h_test, y_h_train, y_h_test = train_test_split(
    X_h, y_h, test_size=0.2, random_state=42, stratify=y_h
)

scaler_h = StandardScaler()
X_h_train_scaled = scaler_h.fit_transform(X_h_train)
X_h_test_scaled = scaler_h.transform(X_h_test)

print("Train shape:", X_h_train.shape)
print("Test shape:", X_h_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

log_model_h = LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42)
log_model_h.fit(X_h_train_scaled, y_h_train)

y_pred_h = log_model_h.predict(X_h_test_scaled)

print("Accuracy:", accuracy_score(y_h_test, y_pred_h))
print("Precision:", precision_score(y_h_test, y_pred_h))
print("Recall:", recall_score(y_h_test, y_pred_h))
print("F1-score:", f1_score(y_h_test, y_pred_h))
print()
print(classification_report(y_h_test, y_pred_h))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model_h = RandomForestClassifier(n_estimators=100, class_weight='balanced', random_state=42, n_jobs=-1)
rf_model_h.fit(X_h_train_scaled, y_h_train)

y_pred_rf_h = rf_model_h.predict(X_h_test_scaled)

print("Accuracy:", accuracy_score(y_h_test, y_pred_rf_h))
print("Precision:", precision_score(y_h_test, y_pred_rf_h))
print("Recall:", recall_score(y_h_test, y_pred_rf_h))
print("F1-score:", f1_score(y_h_test, y_pred_rf_h))
print()
print(classification_report(y_h_test, y_pred_rf_h))
```

```python
from xgboost import XGBClassifier

xgb_model_h = XGBClassifier(
    n_estimators=100,
    scale_pos_weight=(y_h_train.value_counts()[0] / y_h_train.value_counts()[1]),
    random_state=42,
    eval_metric='logloss'
)
xgb_model_h.fit(X_h_train_scaled, y_h_train)

y_pred_xgb_h = xgb_model_h.predict(X_h_test_scaled)

print("Accuracy:", accuracy_score(y_h_test, y_pred_xgb_h))
print("Precision:", precision_score(y_h_test, y_pred_xgb_h))
print("Recall:", recall_score(y_h_test, y_pred_xgb_h))
print("F1-score:", f1_score(y_h_test, y_pred_xgb_h))
print()
print(classification_report(y_h_test, y_pred_xgb_h))
```

```python
import joblib

joblib.dump(log_model_h, '../saved_models/heart_model.pkl')
joblib.dump(scaler_h, '../saved_models/heart_scaler.pkl')

print("Heart disease model and scaler saved!")
```
